### Sentence Transformers for HuggingFace

Hugging Face sentence transformers is Python framework for sentence, text and image embeddings. One of the embedding models is used in HuggingFaceEmbeddings class

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model="all-MiniLM-L6-v2")
```

```python
text="this is a test document"
query_results = embeddings.embed_query(text)
query_results
```

```python
len(query_results)
```

```python
## Multiple query
doc_result = embeddings.embed_documents([text, "this is another test document"])
doc_result
```
